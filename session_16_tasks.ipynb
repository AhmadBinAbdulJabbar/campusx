###`Q-5.`Write a function to change the data type of given a column or a Series. Function takes series and data type as input, returns the converted series.
```
series = pd.Series([1,2,'Python', 2.0, True, 100])
change to float type data
```
Note: Read about pd.to_numeric()

````python
# code here
def change_data_type(series, data_type):
    return pd.to_numeric(series, errors='coerce').astype(data_type)

series = pd.Series([1,2,'Python', 2.0, True, 100])
print(change_data_type(series, float))
````

###`Q-9`
    i. Read `items.csv` making `item_name` as index.
    ii. Show no of nan values
    ii. Item price is given in $, so convert it to rupees without currency symbol.
    iii. Make data type of newly made series as float.
    iv. Fill nan with mean of the series


How csv file looks

```
item_name	item_price
Chips and Fresh Tomato Salsa	$2.39
Izze	$3.39
Nantucket Nectar	$3.39
Chips and Tomatillo-Green Chili Salsa	$2.39
Chicken Bowl	$16.98

```

````python
# code here
items = pd.read_csv('/content/items.csv', index_col='item_name')['item_price']
items.head()
items.isnull().sum()
items = items.str.replace('$', '').astype(float)
items = items.fillna(items.mean())
items.head()
````

###`Q-10`:
    i. Find mean price
    ii. Find 30th and 6th percentile value
    iii. Plot Histogram on price with bin size 50
    iv. No of items price lies between [1000 to 2000]



````python
# code here
print(items.mean())
print(items.quantile(0.3))
print(items.quantile(0.6))
print(items[(items > 1000) & (items < 2000)].count())
items.plot(kind='hist', bins=50)
# items.hist(bins=50)
# items[(items > 1000) & (items < 2000)].count()
# #
````
