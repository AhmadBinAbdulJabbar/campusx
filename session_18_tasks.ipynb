### Handling Missing Values

First, let's identify columns with missing values and their types. Then, for numerical columns, we will fill missing values with the mean of that column.

````python
print('Missing values before filling:')
print(df.isnull().sum()[df.isnull().sum() > 0])

for column in df.columns:
    if df[column].isnull().any():
        if df[column].dtype in ['int64', 'float64']:
            mean_value = df[column].mean()
            df[column].fillna(mean_value, inplace=True)
            print(f"Filled missing values in '{column}' with mean: {mean_value:.2f}")
        else:
            print(f"Column '{column}' has missing values but is not numerical. Skipping mean imputation.")

print('\nMissing values after filling:')
print(df.isnull().sum()[df.isnull().sum() > 0])
````

###`Q-6:` Make a dataframe of each team in IPL with details like - Team Name, Matches Played, Win%, Home Win%, Away Win%.
Show sorted dataframe on Win%

Replace old team name as new name before performing any tasks.
```
Delhi Daredevils ->Delhi Capitals
Kings XI Punjab -> Punjab Kings
Rising Pune Supergiants -> Rising Pune Supergiant
```

Note: Team1 represents Home team. Exclude not result matches.


````python
# code here
````

````python
ipl_cleaned = ipl.copy()

# Replace old team names with new ones
team_name_map = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Rising Pune Supergiants': 'Rising Pune Supergiant'
}

ipl_cleaned['Team1'] = ipl_cleaned['Team1'].replace(team_name_map)
ipl_cleaned['Team2'] = ipl_cleaned['Team2'].replace(team_name_map)
ipl_cleaned['WinningTeam'] = ipl_cleaned['WinningTeam'].replace(team_name_map)

# Exclude 'No Result' matches
ipl_filtered = ipl_cleaned[ipl_cleaned['WinningTeam'] != 'No Result'].copy()
````

````python
team_stats = []
unique_teams = pd.concat([ipl_filtered['Team1'], ipl_filtered['Team2']]).unique()

for team in unique_teams:
    # Matches Played
    matches_played = ipl_filtered[(ipl_filtered['Team1'] == team) | (ipl_filtered['Team2'] == team)].shape[0]

    # Matches Won
    matches_won = ipl_filtered[ipl_filtered['WinningTeam'] == team].shape[0]

    # Home Matches Played (Team1 is home team)
    home_matches_played = ipl_filtered[ipl_filtered['Team1'] == team].shape[0]

    # Home Wins
    home_wins = ipl_filtered[(ipl_filtered['Team1'] == team) & (ipl_filtered['WinningTeam'] == team)].shape[0]

    # Away Matches Played (Team2 is away team)
    away_matches_played = ipl_filtered[ipl_filtered['Team2'] == team].shape[0]

    # Away Wins
    away_wins = ipl_filtered[(ipl_filtered['Team2'] == team) & (ipl_filtered['WinningTeam'] == team)].shape[0]

    # Calculate percentages, handling division by zero
    win_percentage = (matches_won / matches_played * 100) if matches_played > 0 else 0.0
    home_win_percentage = (home_wins / home_matches_played * 100) if home_matches_played > 0 else 0.0
    away_win_percentage = (away_wins / away_matches_played * 100) if away_matches_played > 0 else 0.0

    team_stats.append({
        'Team Name': team,
        'Matches Played': matches_played,
        'Matches Won': matches_won,
        'Win%': round(win_percentage, 2),
        'Home Win%': round(home_win_percentage, 2),
        'Away Win%': round(away_win_percentage, 2)
    })

ipl_team_summary = pd.DataFrame(team_stats)

# Sort by Win% in descending order
ipl_team_summary_sorted = ipl_team_summary.sort_values(by='Win%', ascending=False).reset_index(drop=True)
display(ipl_team_summary_sorted)
````

###`Q-9:` IPL Point Table

Make a function `point_table` which take `season` as parameter and show points table in non-ascendng order of points and in ascending order of team name.

For winning - 2 Ponits;
For loosing - 0 Point
For not result both team gets 1 points.

Make dataframe which will have

````python
def point_table(season):
    # Filter matches for the given season
    season_df = ipl[ipl['Season'] == season].copy()

    # Apply team name mapping to Team1, Team2, and WinningTeam columns
    team_name_map = {
        'Delhi Daredevils': 'Delhi Capitals',
        'Kings XI Punjab': 'Punjab Kings',
        'Rising Pune Supergiants': 'Rising Pune Supergiant'
    }
    season_df['Team1'] = season_df['Team1'].replace(team_name_map)
    season_df['Team2'] = season_df['Team2'].replace(team_name_map)
    season_df['WinningTeam'] = season_df['WinningTeam'].replace(team_name_map)

    # Get all unique teams that played in this season
    all_teams_in_season = pd.concat([season_df['Team1'], season_df['Team2']]).unique()

    # Initialize a dictionary to store stats for each team
    team_stats = {team: {'MatchesPlayed': 0, 'MatchesWon': 0, 'NoResult': 0, 'Points': 0} for team in all_teams_in_season}

    # Iterate through each match to update stats
    for _, row in season_df.iterrows():
        team1 = row['Team1']
        team2 = row['Team2']
        winning_team = row['WinningTeam']

        # Increment matches played for both teams
        team_stats[team1]['MatchesPlayed'] += 1
        team_stats[team2]['MatchesPlayed'] += 1

        if pd.isna(winning_team):  # No Result match
            team_stats[team1]['NoResult'] += 1
            team_stats[team2]['NoResult'] += 1
            team_stats[team1]['Points'] += 1
            team_stats[team2]['Points'] += 1
        else:  # There was a winning team
            team_stats[winning_team]['MatchesWon'] += 1
            team_stats[winning_team]['Points'] += 2

    # Convert the stats dictionary to a DataFrame
    point_table_df = pd.DataFrame.from_dict(team_stats, orient='index')
    point_table_df.index.name = 'TeamName'

    # Sort by Points (non-ascending) and then TeamName (ascending)
    point_table_df = point_table_df.sort_values(by=['Points', 'TeamName'], ascending=[False, True])

    # --- New functionality for SeasonPosition ---
    point_table_with_position = point_table_df.copy()

    # Initialize SeasonPosition with numerical ranks
    point_table_with_position['SeasonPosition'] = range(1, len(point_table_with_position) + 1)
    point_table_with_position['SeasonPosition'] = point_table_with_position['SeasonPosition'].astype(str) # Convert to string to allow 'Winner', 'Runner'

    # Find playoff match results
    final_match_data = season_df[season_df['MatchNumber'] == 'Final']
    qualifier2_match_data = season_df[season_df['MatchNumber'] == 'Qualifier 2']
    eliminator_match_data = season_df[season_df['MatchNumber'] == 'Eliminator']

    # Assign Winner and Runner
    if not final_match_data.empty:
        final_winner = final_match_data.iloc[0]['WinningTeam']
        final_team1 = final_match_data.iloc[0]['Team1']
        final_team2 = final_match_data.iloc[0]['Team2']
        final_runner = final_team1 if final_team1 != final_winner else final_team2

        if final_winner in point_table_with_position.index:
            point_table_with_position.at[final_winner, 'SeasonPosition'] = 'Winner'
        if final_runner in point_table_with_position.index:
            point_table_with_position.at[final_runner, 'SeasonPosition'] = 'Runner'

    # Assign 3rd place (Qualifier 2 loser)
    if not qualifier2_match_data.empty:
        q2_winner = qualifier2_match_data.iloc[0]['WinningTeam']
        q2_team1 = qualifier2_match_data.iloc[0]['Team1']
        q2_team2 = qualifier2_match_data.iloc[0]['Team2']
        q2_loser = q2_team1 if q2_team1 != q2_winner else q2_team2

        if q2_loser in point_table_with_position.index:
            point_table_with_position.at[q2_loser, 'SeasonPosition'] = '3'

    # Assign 4th place (Eliminator loser)
    if not eliminator_match_data.empty:
        eliminator_winner = eliminator_match_data.iloc[0]['WinningTeam']
        eliminator_team1 = eliminator_match_data.iloc[0]['Team1']
        eliminator_team2 = eliminator_match_data.iloc[0]['Team2']
        eliminator_loser = eliminator_team1 if eliminator_team1 != eliminator_winner else eliminator_team2

        if eliminator_loser in point_table_with_position.index:
            point_table_with_position.at[eliminator_loser, 'SeasonPosition'] = '4'

    return point_table_with_position

# Example usage for season 2022
print("IPL Point Table for Season 2022:")
point_table_2022 = point_table('2022')
display(point_table_2022)
````
